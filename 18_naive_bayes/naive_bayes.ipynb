### Training a classifier for continuous features

```python
from sklearn import datasets
from sklearn.naive_bayes import GaussianNB

# Load iris dataset
iris = datasets.load_iris()
features = iris.data
target = iris.target

# Create a Gaussian Naive Bayes classifier
classifier = GaussianNB()

# Train the model
model = classifier.fit(features, target)

# Create a new observation to predict
new_observation = [[4, 4, 4, 0.4]]

# Predict the class of the new observation
print(model.predict(new_observation))
```

### Training a classifier for discrete and quantitative features

```python
import numpy as np
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer

# Создам текстовые данные
text_data = np.array([
    'I love Brazil. Brazil!',
    'Brazil is best',
    'Germany beats both'])

# Create bag of words model
count = CountVectorizer()
bag_of_words = count.fit_transform(text_data)

# Create feature matrix
features = bag_of_words.toarray()

# Create target vector
target = np.array([0,0,1])

# Create a multinomial naive bayes classifier
classifier = MultinomialNB(class_prior=[0.25, 0.5])

# Train the model
model = classifier.fit(features, target)

# Create a new observation
new_observation = [[0, 0, 0, 1, 0, 1, 0]]

# Predict the class of the new observation
print(model.predict(new_observation))
```

### Training a Naive Bayesian Classifier for Binary Features

```python
from sklearn.naive_bayes import BernoulliNB
import numpy as np

# Create a random binary feature matrix
features = np.random.randint(2, size=(100, 3))

# Create a random binary target vector
target = np.random.randint(2, size=(100, 1)).ravel()

# Create object of Bernoulli Naive Bayes classifier
classifer = BernoulliNB(class_prior=[0.25, 0.5])
# Train the model
model = classifer.fit(features, target)
```

### Calibration of predicted probabilities

```python
from sklearn import datasets
from sklearn.naive_bayes import GaussianNB
from sklearn.calibration import CalibratedClassifierCV

# Load iris dataset
iris = datasets.load_iris()
features = iris.data
target = iris.target

# Create a Gaussian Naive Bayes classifier
classifer = GaussianNB()

# Create a sigmoid calibration object based on cross-validation
classifier_sigmoid = CalibratedClassifierCV(classifer, cv=2, method='sigmoid')

# Perform probability calibration
classifier_sigmoid.fit(features, target)

# Create a new observation
new_observation = [[2.6, 2.6, 2.6, 0.4]]

# Output calibrated probability values
classifier_sigmoid.predict_proba(new_observation)
```
