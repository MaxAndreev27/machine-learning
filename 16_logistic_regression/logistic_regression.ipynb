### Training a binary classifier

```python
from sklearn.linear_model import LogisticRegression
from sklearn import datasets
from sklearn.preprocessing import StandardScaler

# Load the iris dataset with two classes
iris = datasets.load_iris()
features = iris.data[:100, :]
target = iris.target[:100]

# Standardize the features
scaler = StandardScaler()
features_standardized = scaler.fit_transform(features)

# Create a logistic regression object
logistic_regression = LogisticRegression(random_state=0)

# Train the model
model = logistic_regression.fit(features_standardized, target)

# Make a prediction for a new observation
new_observation = [[0.5, 0.5, 0.5, 0.5]]

# Predict the class for the new observation
print(model.predict(new_observation))

# Predict the probabilities for the new observation
print(model.predict_proba(new_observation))
```

### Training a multi-class classifier

```python
from sklearn.linear_model import LogisticRegression
from sklearn import datasets
from sklearn.preprocessing import StandardScaler
from sklearn.multiclass import OneVsRestClassifier

# Load the iris dataset
iris = datasets.load_iris()
features = iris.data
target = iris.target

# Standardize the features
scaler = StandardScaler()
features_standardized = scaler.fit_transform(features)

# Create a logistic regression object with one-vs-rest strategy
logistic_regression = LogisticRegression(random_state=0, solver="liblinear")

# Train the model
model = OneVsRestClassifier(logistic_regression).fit(features_standardized, target)
```

### Reducing variance through regularization

```python
from sklearn.linear_model import LogisticRegressionCV
from sklearn import datasets
from sklearn.preprocessing import StandardScaler

# Load the iris dataset
iris = datasets.load_iris()
features = iris.data
target = iris.target

# Standardize the features
scaler = StandardScaler()
features_standardized = scaler.fit_transform(features)

# Create a logistic regression object with cross-validation
logistic_regression = LogisticRegressionCV(
    penalty='l2', Cs=10, random_state=0, n_jobs=-1)

# Train the model
model = logistic_regression.fit(features_standardized, target)
```

### Training a classifier on very large datasets

```python
from sklearn.linear_model import LogisticRegression
from sklearn import datasets
from sklearn.preprocessing import StandardScaler

# Load the iris dataset
iris = datasets.load_iris()
features = iris.data
target = iris.target

# Standardize the features
scaler = StandardScaler()
features_standardized = scaler.fit_transform(features)

# Create a logistic regression
logistic_regression = LogisticRegression(random_state=0, solver="sag")

# Train the model
model = logistic_regression.fit(features_standardized, target)
```

### Managing unbalanced classes

```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn import datasets
from sklearn.preprocessing import StandardScaler

# Load the iris dataset
iris = datasets.load_iris()
features = iris.data
target = iris.target

# Luse balance the dataset by removing the first 40 samples of class 0
features = features[40:, :]
target = target[40:]

# Create a binary target variable (0 for class 1, 1 for class 2)
target = np.where((target == 0), 0, 1)

# Standardize the features
scaler = StandardScaler()
features_standardized = scaler.fit_transform(features)

# Create a logistic regression object with class weights
logistic_regression = LogisticRegression(random_state=0, class_weight="balanced")

# Train the model
model = logistic_regression.fit(features_standardized, target)
```
