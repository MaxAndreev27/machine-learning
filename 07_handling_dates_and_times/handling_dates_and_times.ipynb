### Converting strings to dates

```python
import numpy as np
import pandas as pd
```

```python
date_strings = np.array(
    [
        "03-04-2005 11:35 PM", 
        "23-05-2010 12:01 AM", 
        "04-09-2009 09:09 PM",
        "04-09-2009 09:09 APM"]
)

[pd.to_datetime(date, format='%d-%m-%Y %I:%M %p', errors='coerce') for date in date_strings]
```

### Select date and time

```python
dataframe = pd.DataFrame()
dataframe['date'] = pd.date_range('1/1/2001', periods=100000, freq='h')
filtered_dataframe = dataframe[
    dataframe["date"].between("2002-01-01 01:00:00", "2002-01-01 04:00:00", inclusive="right")
]
filtered_dataframe
```

```python
dataframe = dataframe.set_index(dataframe['date'])
dataframe.loc['2002-1-1 01:00:00':'2002-1-1 04:00:00']
```

### Decomposition of temporal data into individual features

```python
dataframe = pd.DataFrame()
dataframe['date'] = pd.date_range('1/1/2001', periods=150, freq='W')

dataframe['year'] = dataframe['date'].dt.year
dataframe['month'] = dataframe['date'].dt.month
dataframe['day'] = dataframe['date'].dt.day
dataframe['hour'] = dataframe['date'].dt.hour
dataframe['minute'] = dataframe['date'].dt.minute

dataframe.head(10)
```

### Using the sliding window method

```python
time_index = pd.date_range("01/01/2010", periods=5, freq="ME")
dataframe = pd.DataFrame(index=time_index)
dataframe["Stock_Pri.ce"] = [1, 2, 3, 4, 5]
dataframe.rolling(window=2).mean()
```
