```python
import numpy as np
from sklearn.preprocessing import LabelBinarizer, MultiLabelBinarizer
import pandas as pd
```

### Transformation of features with nominal categories

```python
feature = np.array([["Texas"], ["California"], ["Texas"], ["Delaware"], ["Texas"]])
one_hot = LabelBinarizer()
one_hot.fit_transform(feature)
one_hot.classes_
```

#### Reverse transform

```python
one_hot.inverse_transform(one_hot.transform(feature))
```

```python
pd.get_dummies(feature[:, 0])
```

```python
multiclass_feature = [
    ("Texas", "Florida"),
    ("California", "Alabama"),
    ("Texas", "Florida"),
    ("Delaware", "Florida"),
    ("Texas", "Alabama")]

one_hot_multiclass = MultiLabelBinarizer()
one_hot_multiclass.fit_transform(multiclass_feature)
```

```python
one_hot_multiclass.classes_
```

### Transformation of features with ordinal categories

```python
dataframe = pd.DataFrame({"Score": ["Low", "Low", "Medium", "Medium", "High"]})
scale_mapper = {
    "Low": 1,
    "Medium": 2,
    "High": 3}
dataframe["Score"] = dataframe["Score"].replace(scale_mapper)
dataframe.head()
```

```python
dataframe = pd.DataFrame({
    "Score": ["Low", "Low", "Medium", "Medium", "High","Barely More Than Medium"]})
scale_mapper = {
    "Low": 1,
    "Medium": 2,
    "Barely More Than Medium": 3,
    "High": 4}
dataframe["Score"] =dataframe["Score"].replace(scale_mapper)
dataframe.head(6)
```

### Transformation of feature dictionaries

```python
from sklearn.feature_extraction import DictVectorizer

data_dict = [
    {"Red": 2, "Blue": 4},
    {"Red": 4, "Blue": 3},
    {"Red": 1, "Yellow": 2},
    {"Red": 2, "Yellow": 2},
]

dict_vectorizer = DictVectorizer(sparse=False)
data_vectorized = dict_vectorizer.fit_transform(data_dict)
print(data_vectorized)
```
