### Data wrangling

```python
import pandas as pd

url = "https://raw.githubusercontent.com/chrisalbon/sim_data/master/titanic.csv"
df = pd.read_csv(url)
df.head()
```

```python
print("Dataframe shape:", df.shape)
df.describe()
```

```python
df.info()
```

### Show first record

```python
df.iloc[0]
```

### 10 rows and 6 columns

```python
df.iloc[0:10, 0:6]
```

### Min, Max, Mean, Sum, Count

```python
print('Maximum:', df['Age'].max())
print('Minimum:', df['Age'].min())
print('Mean:', df['Age'].mean())
print('Sum:', df['Age'].sum())
print('Count:', df['Age'].count())
```

### Count of value by columns

```python
df.count()
```

### Count of male and female

```python
df['Sex'].value_counts()
```

### Count of people in every class

```python
df['PClass'].value_counts()
```

### Find people without Age value

```python
df[df['Age'].isnull()]
```

### Delete duplicates

```python
df.drop_duplicates()
```

### Count of survived and not

```python
survived_counts = df['Survived'].value_counts().sort_index()

print('Not Survived (0):', survived_counts.get(0, 0))
print('Survived (1):', survived_counts.get(1, 0))
```
