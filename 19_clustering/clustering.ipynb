### K-means clustering

```python
from sklearn import datasets
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Load the iris dataset
iris = datasets.load_iris()
features = iris.data

# Standardize the features
scaler = StandardScaler()
features_std = scaler.fit_transform(features)

# Create a K-means clustering object
cluster = KMeans(n_clusters=3, random_state=0, n_init="auto")

# Fit the model
model = cluster.fit(features_std)

# Output the cluster labels
print("Cluster labels:\n", model.labels_)

# Output the true labels of the iris dataset
print(iris.target)

# Output the cluster centers
print("Cluster centers:\n", model.cluster_centers_)
```

### Acceleration of clustering using the k-means method

```python
from sklearn import datasets
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import MiniBatchKMeans

# Load the iris dataset
iris = datasets.load_iris()
features = iris.data

# Standardize the features
scaler = StandardScaler()
features_std = scaler.fit_transform(features)

# Create a MiniBatch K-means clustering object
cluster = MiniBatchKMeans(n_clusters=3, random_state=0, batch_size=100,
n_init="auto")

# Train the model using MiniBatch K-means
model = cluster.fit(features_std)
```

### Mean-shifted clustering

```python
from sklearn import datasets
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import MeanShift

# Load the iris dataset
iris = datasets.load_iris()
features = iris.data

# Standardize the features
scaler = StandardScaler()
features_std = scaler.fit_transform(features)

# Create a Mean Shift clustering object
cluster = MeanShift(n_jobs=-1)

# Train the model
model = cluster.fit(features_std)
```

### DBSCAN clustering

```python
from sklearn import datasets
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN

# Load the iris dataset
iris = datasets.load_iris()
features = iris.data

# Standardize the features
scaler = StandardScaler()
features_std = scaler.fit_transform(features)

# Create a DBSCAN clustering object
cluster = DBSCAN(n_jobs=-1)

# Train the model
model = cluster.fit(features_std)
```

### Hierarchical clustering

```python
from sklearn import datasets
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering

# Load the iris dataset
iris = datasets.load_iris()
features = iris.data

# Standardize the features
scaler = StandardScaler()
features_std = scaler.fit_transform(features)

# Create an Agglomerative Clustering object
cluster = AgglomerativeClustering(n_clusters=3)

# Train the model
model = cluster.fit(features_std)
```
